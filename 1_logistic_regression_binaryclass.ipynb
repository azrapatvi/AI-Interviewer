```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification

X,y=make_classification(n_samples=1000,n_classes=2,n_features=10,random_state=42)
# 1000 datapoints,10 features or independent features , 2 classes means target column will have two distinct class (binary classification)
# X,y ->dependent and independent features
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
model.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

print(accuracy_score(y_test,y_pred))
print()

print(confusion_matrix(y_test,y_pred))
print()

print(classification_report(y_test,y_pred))
print()
```

#### hyper-parameter tuning and cross validation

##### GridSearchCV

```python
# lasso-> makes the coefficient zero
# ridge-> make the coefficient near to zero

# regularization means try to regularize the patterns in the data...instead of fiting the data

#c -> defines regularization
# c->small c means more regularization : dont try to hard fit (learn)evry single exmaple from the data
# c->large c less regularization : try hard to fit the trainign exampel

params={
    "penalty":['l1','l2','elasticnet'],# l1->lasso l2->ridge
    "C":[0.001,0.01,0.1,1,10],
    "solver":['lbfgs','saga','liblinear','newton-cg']
}

from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold

cv=StratifiedKFold()

grid=GridSearchCV(estimator=model,param_grid=params,cv=cv,scoring='accuracy',n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
```

```python
y_pred
```

```python

print(accuracy_score(y_test,y_pred))
print()

print(confusion_matrix(y_test,y_pred))
print()

print(classification_report(y_test,y_pred))
print()
```
